```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Reload the dataset to ensure we start from the original state
df_titanic = pd.read_csv('titanic.csv')

# Fill missing 'Embarked' values with the mode
df_titanic['Embarked'].fillna(df_titanic['Embarked'].mode()[0], inplace=True)

# Fill missing 'Age' values with the median
df_titanic['Age'].fillna(df_titanic['Age'].median(), inplace=True)

# Remove 'Cabin' column as per user request
df_titanic.drop('Cabin', axis=1, inplace=True)

# Verify that there are no more missing values and show the DataFrame info
display(df_titanic.info())
display(df_titanic.isnull().sum())
```

### Survival Rate by Gender

Let's visualize the survival rate based on gender. This bar plot will show the proportion of males and females who survived.

```python
# Survival Rate by Gender
plt.figure(figsize=(6, 4))
sns.barplot(x='Sex', y='Survived', data=df_titanic, palette='viridis')
plt.title('Survival Rate by Gender')
plt.ylabel('Survival Rate')
plt.xlabel('Gender')
plt.show()
```

### Survival Rate by Passenger Class

Now, let's look at how survival rates differ across passenger classes. This bar plot will display the survival proportion for 1st, 2nd, and 3rd class passengers.

```python
# Survival Rate by Passenger Class
plt.figure(figsize=(7, 5))
sns.barplot(x='Pclass', y='Survived', data=df_titanic, palette='plasma')
plt.title('Survival Rate by Passenger Class')
plt.ylabel('Survival Rate')
plt.xlabel('Passenger Class')
plt.show()
```

### Age Distribution of Survivors vs. Non-Survivors

This visualization will show the age distribution for passengers who survived versus those who did not. A Kernel Density Estimate (KDE) plot is useful here to compare the shapes of the distributions.

```python
# Age Distribution of Survivors vs. Non-Survivors
plt.figure(figsize=(10, 6))
sns.histplot(df_titanic[df_titanic['Survived'] == 1]['Age'], kde=True, color='green', label='Survived', stat='density', alpha=0.5)
sns.histplot(df_titanic[df_titanic['Survived'] == 0]['Age'], kde=True, color='red', label='Did Not Survive', stat='density', alpha=0.5)
plt.title('Age Distribution for Survivors and Non-Survivors')
plt.xlabel('Age')
plt.ylabel('Density')
plt.legend()
plt.show()
```
